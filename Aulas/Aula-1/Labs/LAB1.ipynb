## Como Verificar Consumo de Modelos IA e Quotas no Databricks

Existem várias formas de monitorar o uso de modelos IA no Databricks:

### 1. **Consumo via System Tables** (recomendado)
As tabelas `system.billing.usage` e `system.serving.*` contêm dados detalhados sobre consumo de DBUs e requisições.

### 2. **Quotas e Limites**
No Databricks Community Edition, você tem:
- Limites de compute (15 GB RAM)
- Sem acesso a model serving endpoints pagos
- Acesso limitado a Foundation Models via playground

### 3. **Para contas pagas (AWS/Azure/GCP)**
- Acesse o Account Console → Usage para ver consumo em tempo real
- Configure alertas de quota no Account Console → Quotas
- Use as queries SQL abaixo para análise detalhada

**Nota**: As células SQL abaixo funcionam apenas em workspaces com acesso às System Tables (contas pagas com Unity Catalog habilitado).

```python
%sql
-- Consumo de DBUs por produto IA nos últimos 30 dias
-- (requer Unity Catalog e system.billing.usage habilitado)

SELECT 
  u.usage_date,
  u.sku_name,
  u.usage_unit,
  SUM(u.usage_quantity) as total_usage,
  SUM(u.usage_quantity * list_prices.pricing.default) as estimated_cost_usd
FROM system.billing.usage u
LEFT JOIN system.billing.list_prices ON 
  u.cloud = list_prices.cloud AND
  u.sku_name = list_prices.sku_name AND
  u.usage_date >= list_prices.price_start_time AND
  (u.usage_date < list_prices.price_end_time OR list_prices.price_end_time IS NULL)
WHERE 
  u.usage_date >= CURRENT_DATE - INTERVAL 30 DAYS
  AND (
    u.sku_name LIKE '%MODEL_SERVING%' OR
    u.sku_name LIKE '%AI_GATEWAY%' OR
    u.sku_name LIKE '%SERVERLESS%' OR
    u.sku_name LIKE '%INFERENCE%'
  )
GROUP BY u.usage_date, u.sku_name, u.usage_unit
ORDER BY u.usage_date DESC, total_usage DESC
LIMIT 100;
```

```python
%sql
-- Requisições e métricas dos endpoints de model serving
-- (requer system.serving.* tables)

SELECT 
  e.endpoint_name,
  DATE(u.request_time) as date,
  COUNT(*) as total_requests,
  SUM(u.input_token_count) as total_input_tokens,
  SUM(u.output_token_count) as total_output_tokens,
  AVG(u.input_token_count) as avg_input_tokens,
  AVG(u.output_token_count) as avg_output_tokens,
  SUM(CASE WHEN u.status_code >= 400 THEN 1 ELSE 0 END) as error_count,
  SUM(CASE WHEN u.status_code >= 400 THEN 1 ELSE 0 END) * 100.0 / COUNT(*) as error_rate_pct
FROM system.serving.endpoint_usage u
INNER JOIN system.serving.served_entities e
  ON u.served_entity_id = e.served_entity_id
WHERE u.request_time >= CURRENT_TIMESTAMP - INTERVAL 7 DAYS
GROUP BY e.endpoint_name, DATE(u.request_time)
ORDER BY date DESC, total_requests DESC;
```
